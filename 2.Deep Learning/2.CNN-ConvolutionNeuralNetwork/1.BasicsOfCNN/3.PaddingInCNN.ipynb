## Padding In CNN

#### As we apply filter to image to identify edges , we loose information so we have to do padding to prevent loss of information.
#### So padding means adding 1 more layer of information as a cusion
#### So initial image was 6*6 , and filter was 3*3 , so 4*4 image will be created
## Output pixels = N - F + 1 
- here N - image size , F - filter size 
#### Output = 6 - 3 + 1 = 4
#### But if we add 1 layer of padding on all 4 sides , image will become 8*8
### Output = 8 - 3 + 1 = 6
#### SO we will get 6*6 image as output

## So Padding to be applied is calculated by below formula :

## N - F + 2P + 1 = N
#### 6 - 3 + 2P + 1 = 6
#### 3 + 2P + 1 = 6
#### 2P + 4 = 6
#### 2P = 6 - 4
#### P = 2 / 2 = 1 layer of padding
