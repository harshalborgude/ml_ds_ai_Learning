# Convolution Operation In CNN

### Step 1 : Normalizing
- each pixel value we have to divide by 255
- so if values are 0 and 255 then it will be 0 and 1

### Step 2 : Convolution Operation
- As our visual cortex have various layers V1 , V2 , V3 etc , so that layer work is done by Filters in CNN
- filters can be of different sizes
- so if our image is of 6*6*1 pixels , and filter is of 3*3 then we will place it on top of image starting from pixel 1 and will multiple both pixels to each other and get total sum of it 

- After getting a result we will jump 1 pixel right / left / top / bottom called as stride
#### So stride = 1 if we jump by 1 pixel
- again will do multiplication operation with pixels and sum them
- This way we will keep jumping by stride left , right , up , down untill we conver all pixels
- SO we will get final 4*4 matrix as a result , refer below snapshot.

- Now we can denormalize our final pixel matrix so 0 will become 255 and -4 becomes 0 and will get image as per below
- so just by passing image of 6*6*1 through filter , we can identify the edges.
#### SO this filter is called as vertical edge filter.
#### Same way , horizontal edge filter will be usefull to identify horizontal edges

#### So here image is of size (n) 6 , filter (f) is of size 3 and output we got is of size 4 , so 
### Output Pixel Size = N - F + 1
### Output Pixel Size = 6 - 3 + 1 = 4
#### So here our input was 6*6 and output is of 4*4 so we are loosing information , so to solve this we will use padding.
