# Min , Max & Average Pooling :

### Max Pooling : 
- After applying filters we will get output image , On output image , we will apply Relu activation function after that Max pooling so it will move stride by stride and get max number from the matrix been considered in every stride.
- This concept is used to identify specific feature from the image like , circles , edges etc 
#### This is usefull to identify max intensity pixels.

## Min Pooling :
- The way we are taking maximum element from image in max pooling , similar way in min pooling we will take min number from the image after applying filters & relu activation function.
#### This is usefull to focus on max intensity pixels.


## AVergage Pooling :
- IN average pooling we will get average value from stride of the image after applying filters & relu activation function.
