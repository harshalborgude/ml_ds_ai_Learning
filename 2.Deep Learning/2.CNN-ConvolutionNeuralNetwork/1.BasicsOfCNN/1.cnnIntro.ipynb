# CNN Introduction

- Every image have three chanels (RGB) (Red, Green, Blue)
- Grey images will have only one chanels (Black and White)
- so if a image have 6 * 6 pixels with grey image , it will be 6*6*1 matrix
- In RGB channel , every pixel will be ranging 0-255 
- so all this 3 chanels forms the coloured image.
- so if image is 4*4 pixels then it will be 4*4*3 matrix.
