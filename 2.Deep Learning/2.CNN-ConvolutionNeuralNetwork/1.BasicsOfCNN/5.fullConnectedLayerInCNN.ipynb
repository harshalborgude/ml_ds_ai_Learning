# Fully Connected Layer In CNN :

### Conv_1 : 
- for any image datasets 28*28*1 , lets say we apply valid padding (no padding) and 5*5 filters , then we will get 
- Output pixels = N - F + 1 = 28 - 5 + 1 = 24 * 24 * N1 
- Here we will apply relu activation function 

### Max Pooling (2*2) : 
- After going through max pooling of 2*2 , our image size will decrease to 12*12*n1
- This will be focused on location invariance , most intesity part.

### AGain we will apply convolution operation with no padding and 5*5 Kernal means filter 
### and then again max pooling 
### After this , we will flatten each and every image pixels and will keep attaching it with previous one.
### THis way we will flatten , entire output we got by max pooling.
### This layer is called as fully connected layer , and this is used in ANN now for prediction.
- As in Amnest data set we have images of handwritten 0-9 , so final layer will be having 9 neurons.
### With the forward and backward propogation , we will update filter values 
### IN COnvolution , we apply relu activation function.

# CNN Example with RGB images :

- For RGB image , we use 3 sets of convolutional layers + Relu with pooling , This is called as feature maps.
- So after 3 sets of colvolution + Relu & pooling , we will get flattened layer.
- fully connected neural network will be applied on flattened layer 
- At last we will apply softmax activation function to get multi class classification output like animal names.
- We will try to reduce the output loss with the help of back propagation in training data.
