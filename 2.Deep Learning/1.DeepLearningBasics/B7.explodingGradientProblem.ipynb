# Exploding Gradient Problem : 

- Lets say , while initializing initial weights , we took very hight initial weight 
- so with that other values will be very high bcz of chain rule 
- so bcz of that , W1 new can be very much greather than W1 old.
- Or w1 new can be very less than W1 old 
- so this may lead to rotation of Loss values from here and there on gradient descent curve and it becomes very difficult to achieve global minima , This problem is call exploding gradient problem.
- SO exploding gradient descent problem happens because of wrong weight initialization technique.
