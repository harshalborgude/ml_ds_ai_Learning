# Drop Out Layer :

- overfitting means model performs very high accuracy with training data but comparatively less accuracy with test data called as overfitting.
- Same thing can happen with multilayered neural networks , model performs well with train data but not good with test data called as overfitting of model.
- to avoid overfitting we can use dropout layer.

### SO based on p value (probability) we can decide whether to drop out a particular neuron or not from that perticular layer in forward propogation to avoid overfitting.
### IN test data , we dont dropout neurons with probability , instead we multiply with probability with weights in every layer.
