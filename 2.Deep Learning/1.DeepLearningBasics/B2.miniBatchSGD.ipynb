# Mini Batch Stochastic Gradient Descent (Mini Batch SGD)

### INstead of 1 data point in 1 iteration like stochastic gradient descent , we will have batch-size of data points in 1 iterations.
### Lets say we have 100K data points and batch-size = 1000 , then,
### Iterations in 1 Epoch = 100000 data points / 1000 per batch = 100 iterations

## Advantages :
- 1) noise will reduce compare to SGD as batch size have multiple data points and number of iterations are reduced 
- 2) so convergence speed will increase.
- 3) Efficient resource uses like RAM so time complexity will also improve compare to SGD.

## Disadvantages :
- 1) Noise still exists so still take time to converge.
