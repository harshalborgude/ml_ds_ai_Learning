# Stochastic Gradient Descent 

### In stochastic gradient descent , will provide only 1 record in 1 iteration of forward propogation and backward propogation. So if 1000 data points are there then it takes 1000 iterations to complete 1 Epoch

### will have to do 100s of Epochs to achieve global minima

## Advantages : 
- 1) Solves the resouce issue 

## Disadvantages :
- 1) Time complexity is huge.
- 2) Convergence will take more time.
- 3) Noise gets introduced while will even delay convergence.
