# Gradient Descent Optimizers 

### Optimizers are responsible to reduce the loss function in back propogation by updating the weights.

### using gradient descent our main aim is to update the weights such a way that it reduces loss and reaches global minima on weight to Loss or Cost graph

### Where the slope of Loss is 0 with respect to weight is the global minima point of the graph, Here Wnew == Wold 
- Lets say we have 1000 Data Points , in forward propogation we take 1000 datapoints and compute Yhat that is predicted values for every data point.
- Now we need to reduce this cost function function by using gardient descent optimizers , it will do backward propogation and update weights such that it reduces the loss function.
## So 1 forward propogation with all data points + 1 Backward propogation with all data points = 1 Epoch
- This we will continue doing till we reduce the loss / cost value.
- we can performs 100s of epoc to reduce the cost function .
- with the gradient descent weights will be changing.
## So 1 Epoch = 1 Iteration = 1 Forward Propogation + 1 Backward Propogation
