## Sigmoid Activation Function :

### Derivative of sigmoid will always ranging between 0 to 0.25
### SO multiplication of all sigmoid values will be small value only as each sigmoid value will be ranging between 0 to 0.25 only
### As learning rate will be as small as 0.01 
### So multiplying small value of learning rate with multiplication of all sigmoid will be small value only which will be new W1 new 
### So W1 new is almost simlar to W1 old , so weights will not be updated and we will not move towards convergance and will be stuck at one point only.
### In this case , we are stagnant and thats because of sigmoid activation function's derivative's small value.
### here weights are not updating so this problem called as "Vanishing Gradient Problem" ( W1 new not getting updated compare to W1 old )

## To fix this Vanishing Gradient Problem , Researchers started exploring other activation functions
