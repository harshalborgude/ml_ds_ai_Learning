## Which Activation Function To Use When 

- for hidden layer we cant use Tanh or Sigmoid activation functions as they have vanishing gradient problem 
- so for hidden layers we have to use Relu or its variants - Relu , PreRelu , Leaky Relu , Elu etc
- for output layer and if problem is of binary classification , we can use sigmoid for output layer.
- If its multiclass classification problem where Multiple op neurons are there then use "Softmax Activation Function".
