## Tanh Activation Function :

### tanh(X) activation function transforms output value between -1 to +1 
### derivative of tanh will be ranging between 0 to 1 



## Advantages :
### 1) values are zero centric so weight updation is efficient.


## Disadvantage :
### 1) vanishing gradient problem exist for deep neural network.
### 2) computationally slow as power function is there
