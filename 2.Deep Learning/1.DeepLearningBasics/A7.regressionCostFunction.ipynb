# Loss Functions For Regression Problems :
## 1) Mean Squared Error (MSE) :
### Loss Function = sq( Y - Y hat )
### Cost Function = 1 / n * Sumation (i=1 to n ) * Sq(Yi - Y hat)
### Advantages :
- 1) MSE is differentiable 
- 2) It has 1 local or global minima
- 3) It converges faster 
### Disadvantages :
- 1) Not Robus to outliers (Outliers will have major moment in model)



## 2) Mean Absolute Error (MAE) :
### Loss Function = mod( Y - Y hat )
### Cost Function = 1/n * Sumation(i=1 to n ) * Mod( Yi - Yi hat )
### Advantages :
- 1) Robust to outliers (Moment because of outliers is less)

### Disadvantages "
- Convergence usually takes time in MAE



## 3) Hubur Loss :
- Its combination of MSE & MAE 
- In this , use MSE if hyperparamter is below certain threshold or no outliers otherwise use MAE
