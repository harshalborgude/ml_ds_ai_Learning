## Adagard Optimizers - Adaptive Gradient Descent

### As the convergence happen , the learning rate should change 
### Initially we have high learning rate and then we reduce it as we reach the global minima.

## Formula : Wt new = Wt old - (learning rate dash * derivative of loss function with respect to Wt Old)
- here learning rate is dynamic

## Disadvantages :
- 1) there is possibility of becoming very small value , approx equal to 0 so weight updation will not happen in that case.
