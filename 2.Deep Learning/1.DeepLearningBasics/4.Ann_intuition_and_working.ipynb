## ANN Intuition & Working

## Forward Propogation : 

#### In multi layered neural network , for inital layer values (X1,X2,X3) will be passed to each neuron and weights (w1,w2,w3) will be initialized with bias (b) as below 

### z = sumation(i=1,n) * Xn.Wn + B
- Activation function will be applied to output Z
#### Output of hidden layer 1 will be passed to Hiden Layer 2 , with weight (w4) and bias (b2) , so 
### Z2 = O1*W4 + B2
- Output Z2 will be passed through an activation function
### Activation Func = Activation(Z) = 1 / (1+pov(e) of -z )
- This will be the predicted output of hiden layer 2
### SO the difference between actual output and predicted output is Loss Function
## Loss Function = Y^ - Y
### Based on Loss Function , will have to update all weights (W1,W2,W3,W4) and this process is called as Backward Propogation

## What is Diff Between Loss Function & Cost Function : 
### Loss Function , 
- we are calculation Error
## Pov(Y-Y^) of 2
### Cost Function ,
- We are calculating error for every point
## Sumation(i=1,n) * Pov(Y-Y^) of 2
### So in cost function , we calculate error at once for every point and update the weights at once , unlike Loss Function
