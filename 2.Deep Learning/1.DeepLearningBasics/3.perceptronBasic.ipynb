# What Is Perceptron

- Perceptron is a single layer neural network
- Perceptron is a mathematical model
- for a perceptron , weights and bias are the inputs , and it do summation of all wights and biases , actually its a sumproduct
### Output , Z = W1.X1 + W2.X2 + b
- The Output Z will go to activation function F , and activation function will convert the result based on input, activation functions like step func etc,

 ### z = summation (0 to n) Xi * Wi + b

    - here nose i.e b is like a intercept in term of linear regression 

    - Step 1 : neuron will sum all inputs considering noise and get z value 
    - step 2 : here we apply activation func on z value to transform output

    # Activation function : transformers output between some values of 0 to 1 or -1 to 1 
    - step function : in case of step function if the value is less than equal to 0 , then it will be consider as 0 , 
                        if the value is > 0 then its considered as 1 
    - Sigmoid function : Z > 0.5 then value is 1 
                        if z <= 0.5 then its 0 
                        - here threshold value is 0.5 , this can be changed 
    - why do we apply binary classification ?
        -> to solve any binary classification problem 
    
    - If there is error between actual and predicted output then we have to update weights
    - process of changing weights will go on untill we gets suitable outputs with better accuracy 


# Loss Function :

- SO there are similar 2 process like machine learning
### 1) Training : 
- calculating value of weights and bias is a training so it can be applied for prediction of new data
### 2) Prediction :

## Geometric Intuition :
- Perceptron is actually same like a linior regression model which forms equation based on slope , intercept 
- And based on activation function, it filter outs a region on 2D plane to diferentiate between outputs 
- So Perceptron is a binary classifier which classifies data into 2 different classes by creating 2 regions.
#### region 1 = ( AX + BY + C >= 0 )
#### region 2 = ( AX + BY + C <=> 0)
- same can be done with multiple input features

### In 2D data perceptron will act as a line , and in 3D perceptron will act as a plane and in 4D onwards it will act as a hiperplane
### Perceptron can classify linear or sort of linear data only , it will fail for non linear data

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
## Generating data for cgpa , score & placed / not

np.random.seed(42)  # For reproducibility
cgpa = np.random.uniform(6.0, 10.0, 100).round(2)  # CGPA between 6.0 and 10.0
score = np.random.uniform(50, 100, 100).round(2)   # Score between 50 and 100
placed = ((cgpa > 7.0) & (score > 75)).astype(int) # Placement condition: CGPA > 7.0 and Score > 75

data = {
    'CGPA': list(cgpa),
    'Score': list(score),
    'Placed': list(placed)
}
df = pd.DataFrame(data)
print(df)
```

```python
## Plotting scatter plot using above data

sns.scatterplot(x=df['CGPA'], y=df['Score'], hue=df['Placed'])
plt.title("Scatter Plot of CGPA vs Score with Placement Status")
plt.xlabel("CGPA")
plt.ylabel("Score")
plt.show()
```

```python
## extracting x & y
x = df.iloc[:, 0:2]
y= df.iloc[:, -1]
```

```python
from sklearn.linear_model import Perceptron
p = Perceptron()
```

```python
## Trying to get model out of the data

p.fit(x,y)
```

```python
## Getting the coefficient of model (w1 , w2 in case of perceptron)

p.coef_
```

```python
## getting y axis intercept of model (b i.e bias in case of perceptron)

p.intercept_
```
