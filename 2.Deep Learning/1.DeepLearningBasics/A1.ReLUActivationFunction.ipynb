# ReLU Activation Function  (Rectified Linear Unit)

### Derivative of Relu activation function will be either 0 or 1 in back propogation




### If the derivative of Relu output is 1 , weight updation will happen
### If relu output is 0 , it will create a dead neuron (will pass information as it and no weight updation will happen)

### If z value is +ive then , derivative of Relu will be 1 
### If the z value is -ve then , derivatives of Relu will be 0

## Advantages :
### 1) Solving Vanishing Gradient Problem
### 2) Formula is max(0,1) so calculation is super fast & Relu function has a linear relationsship.
### 3) Much faster than sigmoid / tanH activation function.

## Disadvantages :
### 1) Leading to dead neuron 
### 2) Relu function op is (0,X) so its not zero centric
