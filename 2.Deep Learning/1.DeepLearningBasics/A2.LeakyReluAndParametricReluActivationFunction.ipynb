# Leaky Relu and Parametric Relu Activation Function

### Dead neuron problem of relu activation function is solved by leaky relu & parametric relu
### Function of Leaky relu in forward propogation , 
## max(Alpha * X , X) = max (0.01 * X , X)
### Alpha can be 0.01 etc So we will never get 0 even if value is -ve 
### Output of leaky relu will be always small possitive value for -ve input values 
### As we are not getting 0 value for dead neuron problem gets resolved

## Advantages :
### 1) Leaky relu have all advantages of relu
### 2) It removed the dead relu problem 

## Disadvantages :
### 1) Not Zero centric
